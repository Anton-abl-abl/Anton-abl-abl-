```python
"""

Генерирует моковые отзывы и сохраняет в CSV.
"""

import csv
import random

# --- КОНФИГУРАЦИЯ ---
APP_ID = "com.duolingo.app"
REVIEWS_COUNT = 50
CSV_FILE = "reviews_export.csv"

# Моковые данные
def generate_mock_reviews(count: int):
    languages = ["en", "es", "fr", "de", "it"]
    sample_texts = [
        "Great app, I love it!",
        "This is exactly what I needed.",
        "Fast and reliable.",
        "Some bugs, but overall good.",
        "Not bad, could be better.",
        "Perfect for my daily tasks.",
        "The UI is very intuitive.",
        "I had some issues with login.",
        "Works great on my phone.",
        "Needs more features."
    ]
    reviews = []
    for i in range(count):
        lang = random.choice(languages)
        text = random.choice(sample_texts)
        days_ago = random.randint(0, 30)
        date_str = f"2026-10-{max(1, 26 - days_ago):02d}"
        reviews.append({
            "text": text,
            "date": date_str,
            "language": lang,
            "translation": text  # В этой версии перевод = оригинал
        })
    return reviews

# Сохранение в CSV (с защитой от кодировки)
def save_to_csv(reviews, filename):
    if not reviews:
        print("No reviews to save.")
        return

    fieldnames = ["text", "date", "language", "translation"]
    try:
        with open(filename, mode="w", newline="", encoding="utf-8", errors="replace") as f:
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            for r in reviews:
                writer.writerow(r)
        print(f"✅ Done: CSV saved to {filename}")
    except Exception as e:
        print(f"❌ CSV save failed: {e}")

if __name__ == "__main__":
    reviews_list = generate_mock_reviews(REVIEWS_COUNT)
    save_to_csv(reviews_list, CSV_FILE)
```
